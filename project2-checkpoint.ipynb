```python
import pandas as pd

patients = pd.read_csv("patients_1000.csv")
doctors = pd.read_csv("doctors_1000.csv")
medicines = pd.read_csv("medicines_1000.csv")
prescriptions = pd.read_csv("prescriptions_1000.csv")
```

```python
patients.columns = patients.columns.str.strip().str.lower()
```

```python
patients["gender"] = patients["gender"].fillna("U")
patients["gender"] = patients["gender"].where(
    patients["gender"].isin(["M", "F"]), "U"
)
```

```python
patients = patients[(patients["age"] >= 0) & (patients["age"] <= 100)]
```

```python
patients = patients.drop_duplicates(subset="patient_id")
```

```python
doctors = doctors.drop_duplicates(subset="doctor_id")
medicines = medicines.drop_duplicates(subset="medicine_id")
```

```python
patients
```

```python
doctors
```

```python
prescriptions["dosage"] = prescriptions["dosage"].fillna("Not Provided")
prescriptions["frequency"] = prescriptions["frequency"].fillna("Not Provided")
```

```python

prescriptions = prescriptions.drop_duplicates(subset="prescription_id")
```

```python
prescriptions = prescriptions[
    prescriptions["patient_id"].isin(patients["patient_id"]) &
    prescriptions["doctor_id"].isin(doctors["doctor_id"]) &
    prescriptions["medicine_id"].isin(medicines["medicine_id"])
]
```

```python
from sqlalchemy import create_engine

engine = create_engine(
    "mysql+mysqlconnector://root:2626@localhost/Digital_prescription_db"
)

patients.to_sql("patients", engine, if_exists="append", index=False)
doctors.to_sql("doctors", engine, if_exists="append", index=False)
medicines.to_sql("medicines", engine, if_exists="append", index=False)
prescriptions.to_sql("prescriptions", engine, if_exists="append", index=False)
```

```python
import pandas as pd

query = """ 
SELECT m.medicine_name, COUNT(*) AS total_prescriptions
FROM prescriptions p
JOIN medicines m ON p.medicine_id = m.medicine_id
GROUP BY m.medicine_name
ORDER BY total_prescriptions DESC
LIMIT 10;

"""

df = pd.read_sql(query, engine)
df
```

```python
import matplotlib.pyplot as plt
#MOST FREQUENTLY PRESCRIBED MEDICINES

plt.figure(figsize=(10,5))
plt.bar(df["medicine_name"], df["total_prescriptions"])
plt.xticks(rotation=45)
plt.xlabel("Medicine")
plt.ylabel("Total Prescriptions")
plt.title("Most Frequently Prescribed Medicines")
plt.show()
```

```python
query = """
SELECT d.doctor_name, COUNT(p.prescription_id) AS total_prescriptions
FROM prescriptions p
JOIN doctors d ON p.doctor_id = d.doctor_id
GROUP BY d.doctor_name
LIMIT 12;
"""
#doctor wise prescription
df = pd.read_sql(query, engine)

plt.figure(figsize=(10,5))
plt.bar(df["doctor_name"], df["total_prescriptions"])
plt.xticks(rotation=45)
plt.xlabel("Doctor")
plt.ylabel("Total Prescriptions")
plt.title("Doctor wise Prescription Volume")
plt.show()
```

```python
query = """
SELECT gender, COUNT(*) AS total_prescriptions
FROM patients pt
JOIN prescriptions pr ON pt.patient_id = pr.patient_id
GROUP BY gender;
"""
#Gender wise Prescription Distribution

df = pd.read_sql(query, engine)

plt.figure(figsize=(6,4))
plt.bar(df["gender"], df["total_prescriptions"])
plt.xlabel("Gender")
plt.ylabel("Prescriptions")
plt.title("Gender wise Prescription Distribution")
plt.show()
```

```python
query = """
SELECT
CASE
    WHEN age BETWEEN 0 AND 18 THEN '0-18'
    WHEN age BETWEEN 19 AND 30 THEN '19-30'
    WHEN age BETWEEN 31 AND 45 THEN '31-45'
    WHEN age BETWEEN 46 AND 60 THEN '46-60'
    ELSE '60+'
END AS age_group,
COUNT(pr.prescription_id) AS total_prescriptions
FROM patients pt
JOIN prescriptions pr ON pt.patient_id = pr.patient_id
GROUP BY age_group;
"""

df = pd.read_sql(query, engine)

plt.figure(figsize=(6,4))
plt.bar(df["age_group"], df["total_prescriptions"])
plt.xlabel("Age Group")
plt.ylabel("Prescriptions")
plt.title("Age Group vs Number of Prescriptions")
plt.show()
```

```python
query = """
SELECT doctor_id, COUNT(*) AS total_prescriptions
FROM prescriptions
GROUP BY doctor_id
LIMIT 10;
"""

df = pd.read_sql(query, engine)

plt.figure(figsize=(8,5))
plt.scatter(df["doctor_id"], df["total_prescriptions"])
plt.xlabel("Doctor ID")
plt.ylabel("Prescription Count")
plt.title("Doctors with High Prescription Volume")
plt.show()
```

```python
query = """
SELECT prescribed_date, COUNT(*) AS total_prescriptions
FROM prescriptions
GROUP BY prescribed_date
ORDER BY prescribed_date
LIMIT 10;
"""

df = pd.read_sql(query, engine)

plt.figure(figsize=(10,5))
plt.plot(df["prescribed_date"], df["total_prescriptions"])
plt.xlabel("Date")
plt.ylabel("Prescriptions")
plt.title("Daily Prescription Trend")
plt.xticks(rotation=45)
plt.show()

```
